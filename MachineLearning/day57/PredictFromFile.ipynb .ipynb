```python
import os
import re
import nltk
import joblib
import numpy as np

from sklearn.feature_extraction.text import ENGLISH_STOP_WORDS


MODEL_PATH = r"F:\Data-Science\MachineLearning\day56\sentiment_classifier.joblib"

LABELS = {
    0: "negative",
    1: "positive"
}


class SentimentPredictor:

    def __init__(self, model_path=MODEL_PATH, labels=LABELS):
        self.model_path = model_path
        self.labels = labels
        self.lemmatizer = nltk.stem.WordNetLemmatizer()
        self.classifier = None
        self.vectorizer = None
        self.load_model()

    def text_processor(self, text: str) -> str:
        """Clean and preprocess input text."""

        text = text.lower()

        # Remove HTML tags
        text = re.sub(r"<.+?>", " ", text)

        # Remove URLs
        text = re.sub(r"https?://\S+|www\.\S+", " ", text)

        # Remove punctuation and special characters
        text = re.sub(r"[^\w\s]", " ", text)

        # Remove extra whitespace
        text = re.sub(r"\s+", " ", text).strip()

        # Remove stopwords
        words = [
            word
            for word in text.split()
            if word not in ENGLISH_STOP_WORDS
        ]

        # Remove short words
        words = [
            word
            for word in words
            if len(word) > 2
        ]

        # Lemmatization
        words = [
            self.lemmatizer.lemmatize(word)
            for word in words
        ]

        return " ".join(words)

    def load_model(self):
        """Load classifier and Word2Vec model."""

        if self.model_path is None:
            raise ValueError("model_path is None!")

        if not os.path.exists(self.model_path):
            raise FileNotFoundError(
                f"Model file does not exist: {self.model_path}"
            )

        try:
            package = joblib.load(self.model_path)

            self.classifier = package["classifier"]
            self.vectorizer = package["vectorizer"]

            print("Model loaded successfully.")

        except Exception as e:
            raise RuntimeError(
                f"Cannot load model files: {e}"
            ) from e

    def get_embeddings(self, tokens):
        """Generate mean Word2Vec embedding."""

        vectors = [
            self.vectorizer.wv[word]
            for word in tokens
            if word in self.vectorizer.wv.key_to_index
        ]

        if not vectors:
            raise ValueError(
                "No valid words found in the Word2Vec vocabulary."
            )

        return np.mean(vectors, axis=0)

    def predict(self, user_input: str) -> str:
        """Predict sentiment."""

        processed_text = self.text_processor(user_input)

        if not processed_text:
            raise ValueError(
                "No valid words found after preprocessing."
            )

        tokens = processed_text.split()

        embeddings = np.array([
            self.get_embeddings(tokens)
        ])

        predicted_sentiment = self.classifier.predict(
            embeddings
        )

        return self.labels[
            int(predicted_sentiment[0])
        ].capitalize()

    def run(self, user_input: str):
        """Run prediction."""

        try:
            sentiment = self.predict(user_input)

            print(
                f"Predicted Sentiment: {sentiment}"
            )

        except Exception as e:
            print(
                f"Prediction failed: {e}"
            )


def prediction_pipeline(user_input):
    """Main prediction pipeline."""

    predictor = SentimentPredictor()

    return predictor.predict(user_input)


if __name__ == "__main__":

    predictor = SentimentPredictor()

    user_input = input(
        "Enter your text: "
    )

    predictor.run(user_input)
```
