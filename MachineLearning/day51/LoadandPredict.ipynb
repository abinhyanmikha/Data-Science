```python
import pickle
from tensorflow.keras.models import load_model
import numpy as np
```

```python
model=load_model('iris_classifier.keras')
```

```python
model.summary()
```

```python
with open('iris_classifier.pkl','rb') as file:
    objects=pickle.load(file)
```

```python
scaler=objects['scaler']
encoder=objects['encoder']
pkl_model=objects['model']
```

```python
#user input
my_data=np.array([
    [12.6,2.6,8.5,0.9],
    [6.1,3.0,4.9,1.8]
])
```

```python
classes=encoder.categories_[0]
#preprocessing
scaled_data=scaler.transform(my_data)

#predict
predict=np.argmax(model.predict(scaled_data),axis=1)
[classes[res] for res in predict]
```
