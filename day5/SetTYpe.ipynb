```python
#set operations
a={1,2,3,4,5}
b={4,5,6,7,8}
a.union(b).union({9,10})
a.intersection(b)
a.difference(b)
b.difference(a)
a.intersection_update(b)
a
```

```python
a.symmetric_difference(b)     #a^b
```

```python
a.issubset(b)  #check if a is subset of b
```

```python
b.issuperset(a)  #check if b is superset of a
```
