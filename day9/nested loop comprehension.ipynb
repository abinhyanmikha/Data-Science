```python
# Python
# Input matrix:
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Expected Output: [4, 16, 36, 64]
# Your code here:
squared_evens = [x**2 for matr in matrix for x in matr if x%2==0]
print (squared_evens)
```

```python
# Python
# Input matrix:
matrix = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]
squared_evens =[]
for mat in matrix:
    for i in mat:
        if i%2==0:
            squared_evens.append(i**2)
squared_evens   
```
