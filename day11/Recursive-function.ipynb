```python
def fib(n):
    if n==0:
        return 0
    elif n==1:
        return 1
    else:
        return fib(n-1)+fib(n-2)

fib(8)
```

```python
def factor(n):
    if n==0 or n==1:
     return 1
    return n*factor(n-1)
    
factor(5)
```

```python
cache = [1, 1]

def factorial(n):
    if n < len(cache):
        return cache[n]
    else:
        res = n * factorial(n - 1)
        cache.append(res)
        return res

print(factorial(5))
```
