```python
def get_cube(num):
    return num**3
```

```python
ls=[1,4,8,2,6,9]
```

```python
map_object=map(get_cube,ls)
map_object
```

```python
list(map_object)
```

```python
filter_obj=filter(lambda num: num%2==0,ls)
list(filter_obj)
```
