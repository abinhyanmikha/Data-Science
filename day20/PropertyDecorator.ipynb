```python
import uuid
class Person:
    def __init__(self,name):
        self.__fullname=name
    @property
    def user_id(self):#Getter Method
        print("Getting Name")
        return f"{self.__fullname[-5:]}_{str(uuid.uuid1())[:8]}"
```

Getter method -used to get the value of an attribute. 
called automatically when the attribute is accessed
to define getter method we use @property decorator


```python
obj=Person("Ram Kumar Shrestha")
```

```python
obj.name()
```

```python
obj.name
```

```python
obj.user_id
```
