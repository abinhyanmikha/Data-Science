```python
class product:
    def __init__(self,name,price):
        self.name=name
        self.price=price
```

```python
class user:
    def __init__(self,username,isadmin:bool=False):
        self.username=username
        self.isadmin=isadmin
        self.premium_member=False
        self.cart=shoppingcart(self)#seperate cart for each user gets created automatically
    def set_permium_member(self,admin,status=True):
        if not admin.isadmin:
            print("You are not an admin Cant change the member status")
        else:
            self.premium_member=status
            print(f"{self.username} premium status: {self.premium_member}")
    def create_admin(self,admin,name):
            if not admin.isadmin:
                print("only an admin can create another admin")
                return None
            return  user(name,isadmin=True)

```

```python
class shoppingcart:
    def __init__(self,user):
        self.user=user
        self.cart=[]
    def add_product(self,product):
        self.cart.append(product)
    def remove_product(self,product):
        if product in self.cart:
          self.cart.remove(product)
    def Calculate_total_cost(self):
        total=sum(product.price for product in self.cart)
        if self.user.premium_member==True:
            total=total-0.1*total
        return total
    def generate_invoice(self):
            print("________________INVOICE_______________")
            print(f"Customer name: {self.user.username}")
            print ("\n")
            print("_____________________________________")
            for product in self.cart:
                print(f"| \t {product.name}\t |Rs{product.price}")
            if self.user.premium_member==True:
                print(f"| \t Discount %\t |10%")
            else:
                print(f"| \t Discount %\t |-")
            print("_________________________")
            total = self.Calculate_total_cost()
            print(f"TOTAL \t | Rs {total}")
   
    

    
```
