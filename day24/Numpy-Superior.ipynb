why numpy over for loop?
faster as it performs operation on an entire arrays rather then individual element
for vectorization and caching as lower-level optimization 


```python
import time
import numpy as np
```

```python
a=np.random.rand(90_000_000)
b=np.random.rand(90_000_000)
```

```python
# using core python 
st=time.time()
res=[]
for num1,num2 in zip(list(a),list(b)):
    res.append(num1*num2)
end=time.time()
```

```python
print(f"time taken by python loop: {end-st}")
```

```python
#using list comprehision
st=time.time()
res=[num1*num2 for num1,num2 in zip(list(a),list(b))]
end=time.time()
```

```python
print("time taken by comprehension loop:{end-st}")
```

```python
#using Map
st=time.time()
map_obj=map(lambda a,b:a*b, list(a),list(b))
res=list(map_obj)
end=time.time()
```

```python
print("time taken by comprehension loop:{end-st}")
```
