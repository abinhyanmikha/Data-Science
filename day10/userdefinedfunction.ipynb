```python
#keyword args comes in key value pairs
def print_kwargs(**kwargs):
    print("keyword Args")
    print(kwargs)
print_kwargs(num1=1, num2=2)
type(print_kwargs)
```

```python
def calculate_area(shape:str, **kwargs):
    area=0
    if shape=='square':
        area=kwargs.get("length",0)**2
    elif shape=="rectangle":
        if "length" in kwargs.keys() and "breadth" in kwargs.keys():
            area=kwargs.get("length",0)*kwargs.get("breadth",0)
    else: print(f"{shape} not supported")
    return area
calculate_area("rectangle", length=45, breadth=6)
        
```

```python
def print_args(shape,*args,**kwargs):
    print(shape)
    print(args)
    print(kwargs)
print_args("circle",1,2,3,score=60,greet="hi")
```

Calculate Std. Deviation of given list of numbers.
 
Standard Deviation : $\sqrt{\frac{\sum{(X - \bar{X})}^2}{N}}$
 

```python
from math import sqrt

def calculate_std(*args):
    total = sum(args)
    num = len(args)
    mean = total / num

    std = sqrt(sum((x - mean) ** 2 for x in args) / num)
    return std

print(calculate_std(1, 2, 3))
```
