```python
#args
def print_all(*args):
  print("positional args")
  print(args)
print_all("a",1,"b")
```

```python
def Sum_all(*nums):#*nums is argument
    return sum(nums)
Sum_all(1,2,3,67,8,4,2)
```

```python
#keyword args comes in key value pairs
def print_kwargs(**kwargs):
    print("keyword Args")
    print(kwargs)
print_kwargs(num1=1, num2=2)
type(print_kwargs)
```

```python
def calculate_area(shape:str, **kwargs):
    area=0
    if shape=='square':
        area=kwargs.get("length",0)**2
    elif shape=="rectangle":
        if "length" in kwargs.keys() and "breadth" in kwargs.keys():
            area=kwargs.get("length",0)*kwargs.get("length",0)
    else: print(f"{shape} not supported")
    return area
calculate_area("rectangle", length=45, breadth=6)
        
```
