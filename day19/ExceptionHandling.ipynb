```python
try:
    num=int(input("enter a num"))
    print(6/num)
except ValueError as e:
    print("invalid input")
    print(0)
except Exception as e:
    print("unknown error",e)
    print(0)
print("end of program")
```

```python
# when there is no exeption the else block is executed
try:
    x=float(input("number"))
    y=int(input("number"))
    result=x/y
except ValueError as err:
    print('invalid input', err)
    result=1
else:
    print('final result', result)
```

```python
#finally- always run the code whether an exception occurs or not 
# when there is no exeption the else block is executed
try:
    x=float(input("number"))
    y=int(input("number"))
    result=x/y
except ValueError as err:
    print('invalid input', err)
    result=1
else:
    print('final result', result)
finally:
    print("this is the end of the program")
```
