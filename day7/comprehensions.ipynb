```python
adj=["red", "big", "tasty"]
fruits=["apple", "banana", "cherry"]
#list comprehension

#for a in adj:
 #   for b in fruits:
  #      print(a,b)

pair=[f"{a}-{b}" for a in adj for b in fruits]
pair
```

```python
#dict comprehension
#new dict={key_expression:value_expression for item in iterable if condition==True}
squared={i:i**2 for i in range(1,11)}
squared
```

```python
# Count length of words ( if length > 4)
sent = "Although she was extremely tired after working a long shift at the local hospital, Maria still managed to stop by the grocery store, pick up fresh vegetables for dinner, and help her younger brother with his math homework."
 
# O/P : {words : len(word) } , if length of word is > 4
word={word:len(word) for word in sent.split() if len(word)>4}
word
```

```python
# Count length of words ( if length > 4)
sent = "Although she was extremely tired after working a long shift at the local hospital, Maria still managed to stop by the grocery store, pick up fresh vegetables for dinner, and help her younger brother with his math homework."
 
# O/P : {words : len(word) } , if length of word is > 4
wo={}
for word in sent.split():
    if len(word)>4:
        wo[word]=len(word)
wo
```
