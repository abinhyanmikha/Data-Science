```python
test_numbers = [

    -10, -1, 0, 1, 2, 3, 4, 5, 6, 7, 9, 11, 13, 15, 17, 19, 21, 25, 27, 47, 49, 79, 97, 100, 523, 527, 1000, 7919, 10000, 104729, 1000000

]
def checkprime(num):
    if num < 2:
        return False

    for i in range(2, num):
        if num % i == 0:
            return False

    return True

filtered = filter(checkprime, test_numbers)

print(list(filtered))
```
