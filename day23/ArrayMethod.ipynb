```python
#Concat or Joining
arr1=np.random.randint(1,100, size=(3,3) )
arr2=np.random.randint(1,100, size=(3,3) )
```

```python
arr1, arr2
```

```python
np.concat([arr1,arr2])
```

```python
np.concatenate([arr1,arr2])
```

```python
np.concatenate([arr1,arr2],axis=1)# can use hstack
```

```python
np.concatenate([arr1,arr2],axis=0) # can use vstack
```

```python
np.vstack([arr1,arr2])
```

```python
np.hstack([arr1,arr2])
```
