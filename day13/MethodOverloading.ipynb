```python
class Example:
    def add(self,a,b,c=0):
        return a+b+c
c=Example()
print(c.add(2,4))
print(c.add(2,6,4))
print(c.add(2,3,4.5))
print(c.add("2","3","4"))
```

```python
class Example:
    def add(self,*args):
        return sum(args)
e=Example()
print(e.add(2,4,6,8,0,10,110))
```

```python
#method overriding
class A:
    def pri(self,name):
        self.name=name
        return f"crazy {self.name}" 
class B(A):
    def pri(self,age):
        return f"Jack {age}"
c=B()
print(c.pri("Abi",10))
```
