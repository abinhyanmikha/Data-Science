```python
#mapping type
#represents a collection of key-value pairs, 
# where each key is unique and maps to a specific value. In Python,\
#  the most commonly used mapping type is the dictionary (dict).
#  Dictionaries are mutable, meaning you can add, remove, or change key-value pairs after the dictionary has been created.
```

```python
my_dict={"name":"Abin","age":21, "city":"bhaktapur"}
#key value must be unique, if we use same key again it will overwrite the previous value
my_dict["name"]
my_dict.get("city")#returns the value associated with the key"city"
```
