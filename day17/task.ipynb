```python
class check:
    def __init__(self,passwo:str):
        self.passwo=passwo
    def __str__(self):
        return f"{self.passwo[0]} ***{self.passwo[-1]}"
    def passcheck(self):
        if len(self.passwo)>=8 and self.passwo.isalnum():
            print("strong password")
        else:
            print("weak password")
che=check("checking1")
che.passcheck()

    
```

```python
class check:
    def __init__(self,passwo:str):
        self.passwo=passwo
    def __str__(self):
        return f"{self.passwo[0]} ***{self.passwo[-1]}"
    def passcheck(self):
        if len(self.passwo)>=8 and not self.passwo.isalpha() and not self.passwo.isnumeric():
            print("strong password")
        else:
            print("weak password")
che=check("16565622d32211!!!")
che.passcheck()

    
```
