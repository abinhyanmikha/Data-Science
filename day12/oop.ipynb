```python
class Dog:
    count = 0

    def __init__(self, name: str, age: int = 1):
        self.name = name
        self.age = age
        Dog.count += 1

dog1=Dog("Luffy",3)
dog1.name, dog1.age

```

```python
class Person:
    count=0
    def __init__(self,name:str, address:str="Bhaktapur"):#constructor
        self.name=name
        self.address=address

        Person.count+=1
    def get_details(self):# instance method
        print(f"I am {self.name} from {self.address}")
    @classmethod
    def get_count(cls):#class method
        return cls.count
    @staticmethod
    def get_full_name(first,last):#static method
        return f"{last}, {first}"
student1=Person("Abin")
student1.get_details()
student1.get_count()
```

```python
class Student:
    def __init__(self,name,s1marks,s2marks,s3marks):
        self.name=name
        self.s1marks=s1marks
        self.s2marks=s2marks
        self.s3marks=s3marks
    def total(self):
       return self.s1marks+self.s2marks+self.s3marks
        # return self.tot
    def avg(self):
        return self.total()/3
    
    def display(cls):
        print (f"{cls.name} {cls.s1marks} {cls.s2marks} {cls.s3marks}")
        print(f"total marks is {cls.total()}")
        print(f"avg marks is {cls.avg()}")
student1=Student("Abin",50,50,50)
student1.display()
```
