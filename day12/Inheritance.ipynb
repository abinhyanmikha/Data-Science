```python
class parent:
    def __init__(self,name,age):
        self.name=name
        self.age=age
    def intro(self):
        return f"hello, I am {self.name}"
class child(parent):
    def __init__(self,name1,age1,color):
        super().__init__(name1,age1)
        self.color=color
    def details(self):
        print (f"{self.color} :\n {self.name} \n {self.age}")
child1=child("Ram",15,"green")
child1.intro()

```

```python
class parent1:
    def func1(self):
        print("this is parent 1")
class parent2:
    def func2(self):
        print("this is parent 2")
class parent3:
    def func1(self):
        print("this is parent 3")
class child(parent1,parent2, parent3):
    def func4(self):
        print("this is child function")
obj=child()
obj.func4()
```

```python
class A:
    def method_a(self):
        print("this is method from Class A")
class B(A):
    def method_b(self):
        print("this is method from Class B")
class C(B):
    def method_c(self):
        print("this is method form Class C")
obj=C()
obj.method_b()
```

```python
class Person():
   def __init__(self,name,age):
     self.name=name
     self.age=age
class Employee(Person):
    def __init__(self,name,age,employeeId):
      super.__init__(self,name,age)
      self.employeeId=employeeId
class Manager(Employee):
    def __init__(self, name, age, employeeId,department):
       super().__init__(name, age, employeeId)
       self.department=department
    def display(self):
       return(f"{self.name} is {self.age} years old whose employeeId is {self.employeeId} under {self.department} department")
obj=Manager("Ram",20,1001,"IT")
obj.display()
```
