```python
titanic_df_initials = titanic_df.copy()

titanic_df_initials.initials = titanic_df_initials.initials.replace(
    to_replace={
        "Mlle": "Miss",
        "Mme": "Mrs",
        "Ms": "Mrs",
        "Lady": "Mrs",
        "the Countess": "Mrs",
        "Rev": "Mr",
        "Major": "Mr",
        "Col": "Mr",
        "Don": "Mr",
        "Sir": "Mr",
        "Capt": "Mr",
        "Jonkheer": "Mr",
        
    }
)

titanic_df_initials.initials.value_counts()
```

```python
titanic_df_initials.head(5)
```

```python
titanic_df_initials.head()
```

```python
def replace_with_gender(row):
    if row.Sex=="male":
        value= "Mr"
    value= "Mrs"
    return row.initials.replace("Dr",value,True)
```

```python
titanic_df_initials.initials = titanic_df_initials.apply(replace_with_gender, axis=1)
```
