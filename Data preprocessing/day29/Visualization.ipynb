```python

#Heat map
correlation=titanic_df.corr(numeric_only=True)
correlation
```

```python
#for heat map we need to install seaborn library
import seaborn as sns
sns.heatmap(correlation,annot=True, cmap="coolwarm",fmt=".2f", vmin=-1)
```

```python
import numpy as np
mask=np.triu(np.ones_like(correlation,dtype=bool))
```

```python
sns.heatmap(correlation,annot=True, cmap="coolwarm",fmt=".2f", vmin=-1, mask=mask)
```
