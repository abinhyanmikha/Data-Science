```python
import numpy as np
import pandas as pd
titanic_df=pd.read_csv("titanic_df_Ageoutlier.csv")
```

```python
np.sign(-45)
```

```python
titanic_df.drop("Unnamed: 0",axis=1,inplace=True)
#correlation
titanic_df.corr(numeric_only=True)
```

```python
import scipy.stats as stats
```

```python
titanic_df.head(2)
```

```python
contegency_table=pd.crosstab(titanic_df.PStatus,titanic_df.Survived)
contegency_table
```

```python
chi2_score,pvalue,dof,expected_frequency=stats.chi2_contingency(contegency_table)
```

```python
print(f"Chi-Squared test score: {round(chi2_score,3)}")
print(f"P-value ={round(pvalue,5)}")
```
