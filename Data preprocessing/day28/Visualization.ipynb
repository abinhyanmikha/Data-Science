```python
import pandas as pd
titanic_df=pd.read_csv("F:\github\Data preprocessing\Titanic_data.csv")
```

```python
#Bar Graph
titanic_df.Survived.value_counts().plot(
    kind='bar'
)
```

```python
#Bar Graph
titanic_df.Survived.value_counts().plot(
    kind="bar",
    title="Count of Survived Passangers",
    figsize=(10,7),
    legend=True,
    xlabel="Survived(0=No,1=YEs)",
    ylabel="No of Passangers",
    grid=False, 
    yticks=range(0,551,50),
    colormap="Set1"
)
```

```python
titanic_df.Sex.value_counts().plot(
    kind="bar",
)
```

```python
titanic_df.value_counts(subset=['Sex','Survived']).plot(
    kind="bar",
)
```

```python
pd.crosstab(
    titanic_df.Survived,
    titanic_df.Sex
).plot(
    kind="bar",
    title="Count of Survived Passangers",
    figsize=(10,7),
    legend=True,
    xlabel="Survived(0=No,1=YEs)",
    ylabel="No of Passangers",
    grid=False, 
    yticks=range(0,551,50),
    colormap="Set1"
)
```

```python
ax=titanic_df.Survived.value_counts().plot(
    kind="bar",
    title="Count of Survived Passangers",
    figsize=(10,7),
    legend=True,
    xlabel="Survived(0=No,1=YEs)",
    ylabel="No of Passangers",
    grid=False, 
    yticks=range(0,551,50),
    colormap="Set1"
)
#add number above the bar
ax.bar_label(ax.containers[0],padding=2)
```

Pie Chart

```python
groupby_embarked=titanic_df.groupby(['Embarked'])
groupby_embarked.count()
```

```python
groupby_embarked.count().plot(
    kind="pie",
    y="Survived",
    title="Count of Passangers according to Plass",
    autopct="%1.1f%%",
    startangle=90
)
```

```python
crosstab=pd.crosstab(
    titanic_df.Survived,
    titanic_df.Sex
)
crosstab
```

```python
crosstab.plot(
    kind="pie",
    subplots=True,
    title="Count of Passanger according to Gender and Survived",
)
```

```python
crosstab.T.plot(
    kind="pie",
    subplots=True,
    title="Count of Passanger according to Gender and Survived",
)
```
