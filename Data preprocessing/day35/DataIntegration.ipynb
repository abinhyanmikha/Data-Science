```python
import pandas as pd
df1=pd.DataFrame({
    "acc":[13,21,22,23],"name":["Ram","Sam","hari", "Krishna"],"age":[12,24,32,42]
})
df2=pd.DataFrame({
    "acc":[13,21,22,23],"principle":[500,55,550,600],"interest":[6.3,7.8,9.9,5.5]
})
```

```python
df1
```

```python
df2
```

```python
pd.concat([df1,df2], axis=1)
```

```python
pd.concat([df1,df2], axis=0,ignore_index=True)
```

```python
#Merging
df1, df2
```

```python
pd.merge(df1,df2,how="left",on='acc')
```

```python
df1=pd.DataFrame({
    "acc":[13,20,22,23],"name":["Ram","Sam","hari", "Krishna"],"age":[12,24,32,42]
})
df2=pd.DataFrame({
    "acc":[13,21,22,23,25],"principle":[500,55,550,600,400],"interest":[6.3,7.8,9.9,5.5,2.5]
})
```

```python
pd.merge(df1,df2,how="right",on='acc')
```

```python
pd.merge(df1,df2,how="left",on='acc')
```

```python
pd.merge(df1,df2,how="inner",on='acc',indicator=True)
```

```python
pd.merge(df1,df2,how="outer",on='acc',indicator=True)
```

```python
pd.merge_ordered(df1,df2,how="inner",on='acc')
```

```python
pd.merge_asof(df1,df2,on='acc')
```

```python
df1.merge(df2)
```

```python
#Joining-based on index
df1,df2
```

```python
df2.set_index("acc", inplace=True)
print(df2)
```
