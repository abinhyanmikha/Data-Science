
#DEscription
Analysis of Age and Fare
Age has 714 valid values out of 891 passengers, meaning 177 passengers have missing age data.
The average age is approximately 29.7 years, indicating that the typical passenger was around 30 years old.
The median age is 28 years, which is slightly lower than the mean, suggesting a small influence from older passengers.
The youngest passenger was only 0.42 years old (about 5 months), while the oldest passenger was 80 years old.
The standard deviation of Age is 14.53, showing considerable variation in passengers' ages.
For Fare, all 891 passengers have valid values, so there are no missing Fare values.
The average fare was approximately 32.20, while the median fare was only 14.45.
Since the mean fare (32.20) is much higher than the median fare (14.45), the Fare data is right-skewed, meaning a small number of passengers paid very high fares.
The lowest fare was 0, while the highest fare was 512.33, showing a very large difference between the cheapest and most expensive tickets.
The middle 50% of passengers paid between approximately 7.91 and 31.00.
Overall, Age is relatively spread out, while Fare has strong variation and outliers, particularly among passengers who paid very expensive tickets.
