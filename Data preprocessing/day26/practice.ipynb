```python
df=pd.read_csv("World Largest Cities by Population 2024.csv")
```

```python
#data structure list,dict,tuple and set
df.info()
```

```python
#find cities with population greater than 20 million
new_df=df[df["Population (2024)"]>20000000]["City"]
new_df
```

```python
#find all cities in vetnam
Vietnam=df[df["Country"]=="Vietnam"] 
Vietnam
```

```python
#find cities in china that are growing in population in (2024)
China=df[(df["Country"]=="China") & (df["Growth Rate"]>0)]["City"]
```

```python
China
```

```python
#total popln of largest cities in india

India=df[df["Country"]=="India"]
total_pop=India["Population (2024)"].sum()
total_pop

```

```python
India=df[df["Country"]=="India"]["Population (2024)"].sum()
India
```

```python
#find cities experiencing growth decline and calculate growth median
decline=df[df["Growth Rate"]<0]["Growth Rate"].median()
decline
```

```python
#find growing cities in  China and calculate the minimum population they had in 2023
growing=df[(df["Country"]=="China") &(df["Growth Rate"]>0)]["Population (2023)"].min()
growing
```
