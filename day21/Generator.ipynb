```python
def number_generator(start,end):
    for num in range(start,end+1):
        yield num
```

```python
a=number_generator(4,12)
```

```python
list(a)
```

```python
next(a)
```

```python
for num in a:
    print(num)
```

```python
a=number_generator(10,23)
for num in a:
    print(num)
```

```python
def is_prime(n):
    if n<2:
        return False
    for i in range (2, int(n**0.5)):
        if n%i==0:
            return False
        return True
```

```python
def prime_generator(limit=10):
    count,start=1,1
    while count<=limit:
        start+=1
        if is_prime(start):
            count+=1
            yield start
```

```python
primt_gen_obj=prime_generator(6)
```

```python
list(primt_gen_obj)
```
